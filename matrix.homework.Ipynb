```python
import numpy as np
numbers = [5, 15, 25, 35, 45] 
arr = np.array(numbers)
print(arr)

arr1 = np.zeros(7)
print(arr1)

arr2 = np.ones(6)
print(arr2)

arr3 = np.arange(10, 20)
print(arr3)
```

```python
arr = np.array([12, 24, 43, 48, 60])
arr_multi = arr * 2
print(arr_multi)

arr_add = arr + 10
print(arr_add)

arr_square = arr ** 2
print(arr_square)

arr_sqrt = np.sqrt(arr)
print(arr_square)
```

```python
import numpy as np
data = np.mean([18, 27, 33, 41, 52, 68, 79])
average = np.array(data)
print(average)

data = np.min([18, 27, 33, 41, 52, 68, 79])
min = np.array(data)
print(min)

data = np.max([18, 27, 33, 41, 52, 68, 79])
max = np.array(data)
print(max)

data = np.sum([18, 27, 33, 41, 52, 68, 79])
sum = np.array(data)
print(sum)


               # Вже пізніше вияснив що можна спростити:


data = np.array([18, 27, 33, 41, 52, 68, 79])

average = np.mean(data)
minimum = np.min(data)
maximum = np.max(data)
sum = np.sum(data)

print("Середнє:", average)
print("Мінімум:", minimum)
print("Максимум:", maximum)
print("Сума:", sum)
```

```python
arr = np.array([15, 22, 35, 44, 58, 67])
print("Перший елемент:", arr[0])
print("Останній елемент:", arr[-1])
print("Індекси 1 - 3:", arr[1:4])
print("Елементи > 40:", arr[arr > 40])
```

```python
matrix = np.array([[2, 4, 6],
                  [8, 10, 12],
                  [14, 16, 18]])
print("Форма:", matrix.shape)
print("Перший рядок:", matrix[0])
print("Друга колонка:", matrix[:, 1])
print("Сума:", np.sum(matrix))
print("Середнє:", np.mean(matrix))
```
